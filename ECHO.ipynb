level 2 :The Pre-Agreed Secret

```python
import random
```

```python
def classical_secret_experiment(shots=1000):
  counts = {'00': 0, '11': 0}
  for _ in range(shots):
    secret_bit = random.choice(['0', '1'])
    alice_measurement = secret_bit
    bob_measurement = secret_bit
    outcome = alice_measurement + bob_measurement
    counts[outcome] += 1
  return counts
```

```python
print("Silver Classical Secret Result:", classical_secret_experiment())
```

Output: {'00': \~500, '11': \~500}

Module 05: The Damaged Message

Level 1 :3-Bit Repetition Code &amp; Classical Majority Voting

```python
from qiskit import QuantumCircuit
```

```python
from qiskit_aer import AerSimulator
```

Encoding: Single '1' to 3 Qubits

```python
qc_bronze = QuantumCircuit(3, 3)
```

```python
qc_bronze.x(0)
```

```python
qc_bronze.cx(0, 1)
```

```python
qc_bronze.cx(0, 2)
```

Noise Simulation (Bit-flip on Qubit 1)

```python
qc_bronze.x(1)
```

Measurement

```python
qc_bronze.measure([0, 1, 2], [0, 1, 2])
```

```python
simulator = AerSimulator()
```

```python
counts = simulator.run(qc_bronze, shots=1000).result().get_counts()
```

```python
print("Corrupted Output Signal:", counts) # Expected: {'101': 1000}
```

Python Majority Voting Decoder

```python
raw_output = list(counts.keys())[0]
```

```python
ones_count = raw_output.count('1')
```

```python
decoded_message = '1' if ones_count >= 2 else '0'
```

```python
print(f"Decoded original  Message (Majority Vote): {decoded_message}")
```

* **Bronze:** *"Implements a 3-bit repetition code where a single bit-flip error is successfully recovered using classical majority voting."*
* **Silver:** *"Constructs an in-circuit quantum majority decoder using Toffoli (CCX) gates on an ancilla qubit to automatically correct single bit-flip noise."*
* **Gold:** *"Demonstrates 3-qubit quantum bit-flip error correction, proving that quantum error correction relies on entanglement rather than cloning due to the No-Cloning Theorem."*
